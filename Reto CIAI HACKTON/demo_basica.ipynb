# Demostración Básica - Clasificación de Riesgo Sísmico

Este notebook muestra la clasificación de riesgo sísmico para una zona específica de Santander.

**Nivel Básico:** Cuaderno o interfaz en la que se observe una entrada y el resultado de la clasificación.

## 1. Cargar librerías y datos

```python
import pandas as pd
import numpy as np
from datetime import datetime

# Cargar datos sísmicos
df = pd.read_csv('data/sismic_data_clasificado.csv')
df['fecha'] = pd.to_datetime(df['fecha'])

print(f"Total de eventos sísmicos: {len(df)}")
print(f"\nPrimeros 5 eventos:")
df.head()
```

## 2. Función de clasificación de riesgo

```python
def clasificar_riesgo_sismico(magnitud, profundidad):
    """
    Clasifica el riesgo sísmico basado en magnitud y profundidad.
    
    Parámetros:
    - magnitud: Magnitud del sismo (escala Richter)
    - profundidad: Profundidad del sismo en kilómetros
    
    Retorna:
    - Nivel de riesgo: BAJO, MEDIO, ALTO o CRÍTICO
    """
    if magnitud > 5.5 and profundidad < 50:
        return 'CRÍTICO'
    elif magnitud > 4.5 or profundidad < 50:
        return 'ALTO'
    elif magnitud >= 3.0 or profundidad <= 100:
        return 'MEDIO'
    else:
        return 'BAJO'

print("Función de clasificación cargada correctamente.")
```

## 3. DEMOSTRACIÓN: Clasificar un evento sísmico específico

### INGRESE LOS DATOS DEL SISMO:

```python
# ============================================
# ENTRADA: Modifique estos valores para probar
# ============================================

magnitud_input = 4.2      # Magnitud del sismo (escala Richter)
profundidad_input = 35    # Profundidad en kilómetros
ubicacion_input = "Los Santos"  # Ubicación en Santander

# ============================================
# PROCESAMIENTO
# ============================================

print(f"\n{'='*60}")
print(f"CLASIFICACIÓN DE RIESGO SÍSMICO")
print(f"{'='*60}")
print(f"\n📍 Ubicación: {ubicacion_input}")
print(f"📊 Magnitud: {magnitud_input}")
print(f"🔽 Profundidad: {profundidad_input} km")

# Aplicar clasificación
riesgo = clasificar_riesgo_sismico(magnitud_input, profundidad_input)

# ============================================
# RESULTADO
# ============================================

print(f"\n{'='*60}")
print(f"🔍 RESULTADO DE LA CLASIFICACIÓN")
print(f"{'='*60}")

if riesgo == 'CRÍTICO':
    print(f"\n⚠️  NIVEL DE RIESGO: {riesgo}")
    print(f"   ⚠️  ¡ALERTA MÁXIMA! Evacuación inmediata recomendada.")
elif riesgo == 'ALTO':
    print(f"\n🔴 NIVEL DE RIESGO: {riesgo}")
    print(f"   🚨 Tomar medidas de precaución inmediatas.")
elif riesgo == 'MEDIO':
    print(f"\n🟡 NIVEL DE RIESGO: {riesgo}")
    print(f"   ℹ️  Mantenerse informado y preparado.")
else:
    print(f"\n🟢 NIVEL DE RIESGO: {riesgo}")
    print(f"   ✅ Riesgo bajo, monitoreo normal.")

print(f"\n{'='*60}")
```

## 4. Comparación con datos históricos

```python
# Eventos similares en la base de datos
magnitud_rango = 0.5
profundidad_rango = 20

eventos_similares = df[
    (df['magnitud'].between(magnitud_input - magnitud_rango, magnitud_input + magnitud_rango)) &
    (df['profundidad'].between(profundidad_input - profundidad_rango, profundidad_input + profundidad_rango))
]

print(f"\nEventos históricos similares encontrados: {len(eventos_similares)}")
if len(eventos_similares) > 0:
    print(f"\nDistribución de clasificación de eventos similares:")
    print(eventos_similares['clasificacion'].value_counts())
    print(f"\nÚltimos 5 eventos similares:")
    eventos_similares[['fecha', 'magnitud', 'profundidad', 'ubicacion', 'clasificacion']].tail()
```

## 5. Estadísticas generales de la zona

```python
# Estadísticas por ubicación
stats_por_zona = df.groupby('ubicacion').agg({
    'magnitud': ['mean', 'max', 'count'],
    'profundidad': 'mean',
    'clasificacion': lambda x: (x == 'CRÍTICO').sum()
}).round(2)

stats_por_zona.columns = ['Mag Promedio', 'Mag Máxima', 'Total Eventos', 'Prof Promedio', 'Eventos Críticos']
print("\nEstadísticas por zona en Santander:")
stats_por_zona
```
